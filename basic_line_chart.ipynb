```python
# Using Pandas to plot line graph:
# Row idx will be used for x axis by def (or y axis for horiz bar graph)
# Every col is added to the line graph plot as a line (or other feature)
# The plot type can be specified by passing in 'kind = <argument>'
# Details such as title, axis limits and ticks, figure size can be added.
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

url = 'https://github.com/nytimes/covid-19-data/raw/master/us.csv'
df = pd.read_csv(url)
df.tail()
```

```python
df.plot()  # same as df.plot(kind="line")
```
